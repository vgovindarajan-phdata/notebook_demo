```
import boto3
from botocore.exceptions import ClientError
import _snowflake
print("Completed !!")

aws_sec = _snowflake.get_cloud_provider_token('cred')

def create_s3_client_with_credentials(region_name="us-east-1"):
    # Use AWS Access Key ID and Secret Access Key for authentication
    s3_client = boto3.client(
        's3',
        aws_access_key_id = aws_sec.access_key_id,
        aws_secret_access_key = aws_sec.secret_access_key,
        aws_session_token = aws_sec.token, 
        region_name='us-east-1'
    )
    return s3_client

def list_s3_buckets(s3_client):
    if s3_client:
        try:
            # List all S3 buckets
            response = s3_client.list_buckets()
            print("S3 Buckets:")
            for bucket in response['Buckets']:
                print(f"- {bucket['Name']}")
        except ClientError as e:
            print(f"Error listing S3 buckets: {e}")
            
# Example usage
s3_client = create_s3_client_with_credentials(region_name="us-east-1")
list_s3_buckets(s3_client)
```
